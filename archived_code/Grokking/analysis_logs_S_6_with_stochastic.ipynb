```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    'font.size': 12,
    'font.family': 'serif',
    'axes.grid': True,
    'grid.alpha': 0.4,
    'grid.linestyle': '--'
})

csv_path = 'grokking_logs/grokking_modular_addition_logs_S_5_with_stochastic.csv'
df = pd.read_csv(csv_path)

steps = pd.to_numeric(df['step'], errors='coerce').values
train_loss = pd.to_numeric(df['train_loss'], errors='coerce').values
val_loss = pd.to_numeric(df['val_loss'], errors='coerce').values

window = max(1, len(steps) // 100)
train_loss_smooth = pd.Series(train_loss).rolling(window=window, min_periods=1).mean().values
val_loss_smooth = pd.Series(val_loss).rolling(window=window, min_periods=1).mean().values

fig, ax = plt.subplots(figsize=(12, 6))

ax.plot(steps, train_loss, color='tab:blue', alpha=0.15)
ax.plot(steps, val_loss, color='tab:red', alpha=0.15)

ax.plot(steps, train_loss_smooth, color='tab:blue', linewidth=2.5, label='Train Loss (Smoothed)')
ax.plot(steps, val_loss_smooth, color='tab:red', linewidth=2.5, label='Val Loss (Smoothed)')

ax.set_yscale('log')
ax.set_xlabel('Optimization Steps', fontsize=14, fontweight='bold')
ax.set_ylabel('Cross-Entropy Loss (Log Scale)', fontsize=14, fontweight='bold')

ax.set_xlim(steps.min(), steps.max())
ax.set_ylim(bottom=max(1e-7, min(train_loss.min(), val_loss.min()) * 0.5))

ax.set_title('Training Dynamics on $S_5$ Composition\n(Stochastic Mini-Batch Training, WD = 1)', fontsize=15, pad=15)
ax.legend(loc='upper right', framealpha=0.9, fontsize=12)

plt.tight_layout()
plt.savefig('s5_with_stochastic_loss_dynamics.pdf', format='pdf', dpi=300)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Настройка академического стиля графиков
plt.rcParams.update({
    'font.size': 12,
    'font.family': 'serif',
    'axes.grid': True,
    'grid.alpha': 0.4,
    'grid.linestyle': '--'
})

csv_path = 'grokking_logs/grokking_modular_addition_logs_S_5_with_stochastic.csv'
df = pd.read_csv(csv_path)

# Убедимся, что данные числовые
steps = pd.to_numeric(df['step'], errors='coerce').values
train_loss = pd.to_numeric(df['train_loss'], errors='coerce').values
val_loss = pd.to_numeric(df['val_loss'], errors='coerce').values

# 2. Вычисляем скользящее среднее (сглаживание шума SGD)
window = max(1, len(steps) // 100) # Автоматический размер окна (1% от длины ряда)
train_loss_smooth = pd.Series(train_loss).rolling(window=window, min_periods=1).mean().values
val_loss_smooth = pd.Series(val_loss).rolling(window=window, min_periods=1).mean().values

# 3. Создаем график
fig, ax = plt.subplots(figsize=(12, 6))

# Рисуем СЫРЫЕ данные полупрозрачным фоном (показывают реальный хаос системы)
ax.plot(steps, train_loss, color='tab:blue', alpha=0.15)
ax.plot(steps, val_loss, color='tab:red', alpha=0.15)

# Рисуем СГЛАЖЕННЫЕ данные толстыми линиями (показывают макро-динамику)
ax.plot(steps, train_loss_smooth, color='tab:blue', linewidth=2.5, label='Train Loss (Smoothed)')
ax.plot(steps, val_loss_smooth, color='tab:red', linewidth=2.5, label='Val Loss (Smoothed)')

# 4. Настройка осей (ОБЫЧНАЯ / ЛИНЕЙНАЯ ШКАЛА)
ax.set_xlabel('Optimization Steps', fontsize=14, fontweight='bold')
ax.set_ylabel('Cross-Entropy Loss', fontsize=14, fontweight='bold')

# Настраиваем границы X
ax.set_xlim(steps.min(), steps.max())

# Настраиваем границы Y для линейного графика (от нуля с небольшим запасом сверху)
max_y = max(train_loss_smooth.max(), val_loss_smooth.max())
ax.set_ylim(bottom=-0.1, top=max_y * 1.1)

# 5. Оформление
ax.set_title('Training Dynamics on $S_5$ Composition\n(Linear Scale, Stochastic Mini-Batch, WD = 1)', fontsize=15, pad=15)
ax.legend(loc='upper right', framealpha=0.9, fontsize=12)

# Сохраняем в векторном формате для статьи
plt.tight_layout()
plt.savefig('s5_with_stochastic_loss_dynamics_linear.pdf', format='pdf', dpi=300)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tqdm.auto import tqdm
from search_for_optimal_parameters import mle_intrinsic_dimension

plt.rcParams.update({
    'font.size': 14,
    'font.family': 'serif',
    'axes.grid': True,
    'grid.alpha': 0.4,
    'grid.linestyle': '--'
})

def generate_grokking_presentation_plots(
    csv_path, 
    window_size=300, 
    step_size=50, 
    prefix="mod_wd1"
):
    df = pd.read_csv(csv_path)
    
    for col in['step', 'train_loss', 'val_loss', 'val_acc', 'weight_norm']:
        df[col] = pd.to_numeric(df[col], errors='coerce')
    df = df.dropna(subset=['step', 'val_acc', 'weight_norm']).reset_index(drop=True)
    
    all_steps = np.array(df['step'].values, dtype=np.float64)
    
    series = df['weight_norm'].values.astype(np.float64)
    
    steps_E =[]
    E_history =[]
    
    for i in tqdm(range(0, len(series) - window_size + 1, step_size)):
        window_data = series[i : i + window_size]
        
        if np.isnan(window_data).any() or len(window_data) < window_size:
            continue
            
        E = mle_intrinsic_dimension(window_data, tau=1, max_E=15, k_neighbors=5)
        
        center_step = all_steps[i + window_size // 2]
        
        if E is not None and not np.isnan(E) and not np.isinf(E):
            clean_E = max(1.0, min(float(E), 30.0))
        else:
            clean_E = np.nan
            
        steps_E.append(center_step)
        E_history.append(clean_E)

    x_e = np.array(steps_E, dtype=np.float64)
    y_e = np.array(E_history, dtype=np.float64)
    x_full = all_steps

    grok_step = None
    grok_df = df[df['val_acc'] >= 0.95]
    if not grok_df.empty:
        grok_step = grok_df['step'].iloc[0]

    color_E = 'tab:purple'
    
    fig1, ax1 = plt.subplots(figsize=(9, 5))
    
    ax1.set_xlabel('Optimization Steps', fontsize=14, fontweight='bold')
    ax1.set_ylabel('Dimensionality $E$', color=color_E, fontsize=14, fontweight='bold')
    line_E1, = ax1.plot(x_e, y_e, color=color_E, linewidth=3, marker='o', markersize=4, label='$E$ (MLE)')
    ax1.tick_params(axis='y', labelcolor=color_E)
    
    valid_E = y_e[~np.isnan(y_e)]
    if len(valid_E) > 0:
        ax1.set_ylim(min(valid_E) - 0.05, max(valid_E) + 0.05)

    ax2 = ax1.twinx()
    ax2.set_ylabel('Loss', color='black', fontsize=14, fontweight='bold')
    line_tr, = ax2.plot(x_full, df['train_loss'], color='tab:blue', alpha=0.3, label='Train Loss')
    line_val, = ax2.plot(x_full, df['val_loss'], color='tab:red', linewidth=2.5, label='Val Loss')
    ax2.tick_params(axis='y', labelcolor='black')

    ax1.set_xlim(x_full.min(), x_full.max())

    if grok_step:
        ax2.axvline(x=grok_step, color='black', linestyle=':', linewidth=2)
        text_x = min(grok_step + 300, x_full.max() - 2000)
        y_pos = df['val_loss'].max() * 0.5
        ax2.annotate('Grokking', xy=(grok_step, y_pos), xytext=(text_x, y_pos),
                     arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=6),
                     fontsize=12, fontweight='bold')

    lines1 = [line_E1, line_tr, line_val]
    labels1 =[l.get_label() for l in lines1]
    ax1.legend(lines1, labels1, loc='center left', framealpha=0.9, fontsize=11)
    
    plt.tight_layout()
    path_loss = f'{prefix}_loss.png'
    plt.savefig(path_loss, format='png', dpi=300, bbox_inches='tight')
    plt.close(fig1)

    fig2, ax3 = plt.subplots(figsize=(9, 5))
    
    ax3.set_xlabel('Optimization Steps', fontsize=14, fontweight='bold')
    ax3.set_ylabel('Dimensionality $E$', color=color_E, fontsize=14, fontweight='bold')
    line_E2, = ax3.plot(x_e, y_e, color=color_E, linewidth=3, marker='o', markersize=4, label='$E$ (MLE)')
    ax3.tick_params(axis='y', labelcolor=color_E)
    
    if len(valid_E) > 0:
        ax3.set_ylim(min(valid_E) - 0.05, max(valid_E) + 0.05)

    ax4 = ax3.twinx()
    color_norm = 'tab:green'
    ax4.set_ylabel('Weight Norm', color=color_norm, fontsize=14, fontweight='bold')
    line_norm, = ax4.plot(x_full, df['weight_norm'], color=color_norm, linewidth=2.5, linestyle='--', label='Weight Norm')
    ax4.tick_params(axis='y', labelcolor=color_norm)
    
    norm_min, norm_max = df['weight_norm'].min(), df['weight_norm'].max()
    ax4.set_ylim(norm_min * 0.9, norm_max * 1.1)

    ax3.set_xlim(x_full.min(), x_full.max())

    if grok_step:
        y_mid = (norm_max + norm_min) / 2
        ax4.axvline(x=grok_step, color='black', linestyle=':', linewidth=2)
        ax4.annotate('Grokking', xy=(grok_step, y_mid), xytext=(text_x, y_mid),
                     arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=6),
                     fontsize=12, fontweight='bold')

    lines2 = [line_E2, line_norm]
    labels2 =[l.get_label() for l in lines2]
    ax3.legend(lines2, labels2, loc='center left', framealpha=0.9, fontsize=11)
    
    plt.tight_layout()
    path_norm = f'{prefix}_norm.png'
    plt.savefig(path_norm, format='png', dpi=300, bbox_inches='tight')
    plt.close(fig2)

    fig3, ax5 = plt.subplots(figsize=(9, 5))
    
    ax5.set_xlabel('Optimization Steps', fontsize=14, fontweight='bold')
    ax5.set_ylabel('Dimensionality $E$', color=color_E, fontsize=14, fontweight='bold')
    line_E3, = ax5.plot(x_e, y_e, color=color_E, linewidth=3, marker='o', markersize=4, label='$E$ (MLE)')
    ax5.tick_params(axis='y', labelcolor=color_E)
    
    if len(valid_E) > 0:
        ax5.set_ylim(min(valid_E) - 0.05, max(valid_E) + 0.05)

    ax6 = ax5.twinx()
    color_tr_acc = 'tab:blue'
    color_val_acc = 'tab:red'
    ax6.set_ylabel('Accuracy', color='black', fontsize=14, fontweight='bold')
    line_tr_acc, = ax6.plot(x_full, df['train_acc'], color=color_tr_acc, alpha=0.3, label='Train Acc')
    line_val_acc, = ax6.plot(x_full, df['val_acc'], color=color_val_acc, linewidth=2.5, label='Val Acc')
    ax6.tick_params(axis='y', labelcolor='black')
    ax6.set_ylim(-0.05, 1.05)

    ax5.set_xlim(x_full.min(), x_full.max())

    if grok_step:
        ax6.axvline(x=grok_step, color='black', linestyle=':', linewidth=2)
        ax6.annotate('Grokking', xy=(grok_step, 0.2), xytext=(text_x, 0.2),
                     arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=6),
                     fontsize=12, fontweight='bold')

    lines3 =[line_E3, line_tr_acc, line_val_acc]
    labels3 = [l.get_label() for l in lines3]
    ax5.legend(lines3, labels3, loc='center left', framealpha=0.9, fontsize=11)
    
    plt.tight_layout()
    path_acc = f'{prefix}_acc.png'
    plt.savefig(path_acc, format='png', dpi=300, bbox_inches='tight')
    plt.close(fig3)


csv_s5_wd1 = 'grokking_logs/grokking_modular_addition_logs_S_5_with_stochastic.csv'
generate_grokking_presentation_plots(
    csv_s5_wd1, 
    window_size=300, 
    step_size=50, 
    prefix="s5_wd1"
)

csv_s5_wd0 = 'grokking_logs/grokking_modular_addition_logs_S_5_with_stochastic_without_weight_decay.csv'
generate_grokking_presentation_plots(
    csv_s5_wd0, 
    window_size=300, 
    step_size=50, 
    prefix="s5_wd0"
)
```
